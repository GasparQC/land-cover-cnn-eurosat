# 2. Descripción completa del problema a resolver

El problema elegido consiste en desarrollar un modelo de clasificación de imágenes capaz de reconocer automáticamente distintos tipos de cobertura territorial a partir de imágenes satelitales.

El modelo deberá recibir como entrada una imagen satelital y clasificarla en una de las 10 categorías del dataset EuroSAT. Estas categorías representan distintos usos o coberturas del suelo, como zonas agrícolas, áreas forestales, autopistas, zonas industriales, áreas residenciales, ríos, lagos o mares.

Este problema pertenece al campo de la **visión por computadora** y de la **teledetección**. Su resolución mediante redes neuronales convolucionales resulta adecuada porque las CNN pueden aprender patrones visuales presentes en las imágenes, tales como texturas, formas, bordes, colores, densidad de elementos, trazados urbanos y estructuras territoriales.

A diferencia de una fotografía tomada desde una perspectiva frontal, una imagen satelital muestra el territorio desde una perspectiva cenital. Esto implica que el modelo no debe reconocer objetos aislados, sino configuraciones espaciales, patrones de distribución y relaciones visuales entre superficies.

Por ejemplo, una zona residencial puede distinguirse por la repetición de techos y calles; una autopista por trazados lineales; un bosque por texturas vegetales densas; un área agrícola por patrones geométricos de cultivo; y un río por formas alargadas e irregulares asociadas al agua.

El desafío principal del problema consiste en que algunas clases pueden compartir características visuales similares. Las clases agrícolas y vegetales pueden tener colores y texturas cercanas; las clases urbanas pueden compartir superficies grises o formas rectangulares; y las clases vinculadas al agua pueden presentar tonos y estructuras visuales parecidas. Por este motivo, el modelo deberá aprender diferencias visuales generales y también rasgos más finos entre categorías.

# 6. EDA - Análisis exploratorio inicial de datos

El análisis exploratorio de datos permite conocer la estructura del dataset antes de entrenar el modelo. En esta etapa se busca:

Para realizar este EDA se trabajará con el archivo comprimido del dataset, subido manualmente a Google Colab. El notebook descomprime el archivo ZIP, identifica automáticamente la carpeta que contiene las clases y construye un DataFrame con la ruta de cada imagen y su etiqueta.

```python
# Importación de librerías principales

import os
import zipfile
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
```

```python
# Subir el archivo ZIP del dataset desde la computadora

from google.colab import files

uploaded = files.upload()
```

```python
# Detectar automáticamente el archivo .zip subido

zip_files = [nombre for nombre in uploaded.keys() if nombre.lower().endswith(".zip")]

if len(zip_files) == 0:
    raise FileNotFoundError("No se subió ningún archivo .zip.")

ZIP_PATH = zip_files[0]

print("Archivo ZIP detectado:", ZIP_PATH)
```

```python
# Descomprimir el ZIP en una carpeta de trabajo

EXTRACT_PATH = "/content/eurosat_dataset"

os.makedirs(EXTRACT_PATH, exist_ok=True)

with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
    zip_ref.extractall(EXTRACT_PATH)

print("Dataset descomprimido en:", EXTRACT_PATH)
```

```python
# Revisar carpetas creadas después de descomprimir

for root, dirs, files in os.walk(EXTRACT_PATH):
    print("Carpeta:", root)
    print("Subcarpetas:", dirs[:10])
    print("Cantidad de archivos:", len(files))
    print("-" * 60)

    # Solo mostramos los primeros niveles para no imprimir demasiado
    if root.count(os.sep) >= EXTRACT_PATH.count(os.sep) + 2:
        dirs[:] = []
```

```python
# Buscar automáticamente la carpeta que contiene las clases del dataset

CLASES_ESPERADAS = {
    "AnnualCrop",
    "Forest",
    "HerbaceousVegetation",
    "Highway",
    "Industrial",
    "Pasture",
    "PermanentCrop",
    "Residential",
    "River",
    "SeaLake"
}

DATASET_PATH = None

for root, dirs, files in os.walk(EXTRACT_PATH):
    dirs_set = set(dirs)
    if CLASES_ESPERADAS.issubset(dirs_set):
        DATASET_PATH = root
        break

if DATASET_PATH is None:
    raise FileNotFoundError(
        "No se encontró una carpeta que contenga las 10 clases esperadas. "
        "Revisar la estructura del ZIP."
    )

print("Carpeta del dataset encontrada:", DATASET_PATH)
```

```python
# Identificar las clases a partir de las carpetas

clases = sorted([
    carpeta for carpeta in os.listdir(DATASET_PATH)
    if os.path.isdir(os.path.join(DATASET_PATH, carpeta))
])

print("Cantidad de clases encontradas:", len(clases))
print("Clases:")
for clase in clases:
    print("-", clase)
```

```python
# Crear un DataFrame con la ruta de cada imagen y su clase

formatos_validos = (".jpg", ".jpeg", ".png", ".tif", ".tiff")

data = []

for clase in clases:
    carpeta_clase = os.path.join(DATASET_PATH, clase)

    for nombre_archivo in os.listdir(carpeta_clase):
        if nombre_archivo.lower().endswith(formatos_validos):
            ruta = os.path.join(carpeta_clase, nombre_archivo)

            data.append({
                "filepath": ruta,
                "clase": clase,
                "archivo": nombre_archivo
            })

df = pd.DataFrame(data)

df.head()
```

```python
# Información general del dataset organizado en DataFrame

print("Cantidad total de imágenes encontradas:", len(df))
print("Cantidad de clases encontradas:", df["clase"].nunique())

df.info()
```

```python
# Verificar si hay rutas duplicadas

duplicados = df["filepath"].duplicated().sum()

print("Cantidad de rutas duplicadas:", duplicados)
```

```python
# Conteo de imágenes por clase

conteo_clases = df["clase"].value_counts().sort_index()

df_conteo = conteo_clases.reset_index()
df_conteo.columns = ["Clase", "Cantidad de imágenes"]

df_conteo
```

```python
# Agregar porcentaje de imágenes por clase

df_conteo["Porcentaje"] = (df_conteo["Cantidad de imágenes"] / len(df)) * 100
df_conteo
```

```python
# Gráfico de distribución de imágenes por clase

plt.figure(figsize=(10, 5))
plt.bar(df_conteo["Clase"], df_conteo["Cantidad de imágenes"])
plt.title("Distribución de imágenes por clase - EuroSAT RGB")
plt.xlabel("Clase")
plt.ylabel("Cantidad de imágenes")
plt.xticks(rotation=45, ha="right")
plt.show()
```

# 7. Análisis visual cuantitativo básico

Además de observar ejemplos, se realiza un análisis visual cuantitativo simple sobre una muestra del dataset. Este análisis no entrena ningún modelo, pero permite describir rasgos generales de las imágenes.

Se calcularán:

- brillo promedio por imagen;
- promedio de canales RGB por imagen;
- estadísticas por clase;
- imagen promedio por clase.

Estos valores ayudan a comprender si existen diferencias visuales generales entre las categorías.

```python
# Crear una muestra para análisis de color y brillo
# Se usa una muestra para que el análisis sea rápido en Colab.

MUESTRA_ANALISIS = min(3000, len(df))

df_muestra_color = df.sample(MUESTRA_ANALISIS, random_state=42).reset_index(drop=True)

print("Cantidad de imágenes utilizadas para análisis de color:", len(df_muestra_color))
```

```python
# Calcular brillo promedio y promedio de canales RGB por imagen

datos_color = []

for fila in df_muestra_color.itertuples():
    imagen = Image.open(fila.filepath).convert("RGB")
    arr = np.array(imagen)

    brillo_promedio = arr.mean()
    r_promedio = arr[:, :, 0].mean()
    g_promedio = arr[:, :, 1].mean()
    b_promedio = arr[:, :, 2].mean()

    datos_color.append({
        "clase": fila.clase,
        "brillo_promedio": brillo_promedio,
        "R_promedio": r_promedio,
        "G_promedio": g_promedio,
        "B_promedio": b_promedio
    })

df_color = pd.DataFrame(datos_color)

df_color.head()
```

```python
# Estadísticas de brillo por clase

brillo_por_clase = (
    df_color
    .groupby("clase")["brillo_promedio"]
    .agg(["mean", "std", "min", "max"])
    .reset_index()
)

brillo_por_clase
```

```python
# Gráfico de brillo promedio por clase

plt.figure(figsize=(10, 5))
plt.bar(brillo_por_clase["clase"], brillo_por_clase["mean"])
plt.title("Brillo promedio por clase - EuroSAT RGB")
plt.xlabel("Clase")
plt.ylabel("Brillo promedio")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## Interpretación del brillo promedio por clase

El gráfico de brillo promedio muestra diferencias generales de luminosidad entre las clases del dataset.

Las clases con mayor brillo promedio son **Industrial**, **AnnualCrop** y **PermanentCrop**. Esto puede relacionarse con la presencia de superficies claras, áreas agrícolas expuestas, caminos, techos o estructuras artificiales que reflejan más luz en la imagen.

```python
# Crear una imagen promedio por clase usando una muestra de imágenes

imagenes_promedio = {}

cantidad_por_clase = 100  # Se usa un número bajo para que no tarde demasiado

for clase in clases:
    rutas_clase = df[df["clase"] == clase]["filepath"].sample(
        min(cantidad_por_clase, len(df[df["clase"] == clase])),
        random_state=42
    )

    acumulador = np.zeros((64, 64, 3), dtype=np.float32)

    for ruta in rutas_clase:
        imagen = Image.open(ruta).convert("RGB")
        arr = np.array(imagen).astype(np.float32)
        acumulador += arr

    promedio = acumulador / len(rutas_clase)
    imagenes_promedio[clase] = promedio.astype(np.uint8)
```

```python
# Mostrar imágenes promedio por clase

plt.figure(figsize=(14, 8))

for i, clase in enumerate(clases):
    plt.subplot(2, 5, i + 1)
    plt.imshow(imagenes_promedio[clase])
    plt.title(clase)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

Las clases **Industrial** y **Residential** presentan promedios más claros y grisáceos, lo que puede relacionarse con superficies construidas, techos, calles y estructuras urbanas. **River** presenta una tonalidad intermedia, ya que sus imágenes pueden incluir agua, vegetación y bordes territoriales.
